**Archivos que debes producir:** `app.py`


### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="Tu servidor responderá exactamente así")

# Modelo de validación para el mensaje del eco
class MensajeEntrada(BaseModel):
    mensaje: str

# Ruta 1: Estado del servidor
@app.get("/api/estado", status_code=200)
def obtener_estado():
    return {"servidor": "ok", "curso": "DAB-2201"}

# Ruta 2: Eco
@app.post("/api/eco", status_code=200)
def eco(mensaje: MensajeEntrada):
    return {
        "mensaje": mensaje.mensaje,
        "recibido": True
    }
````
